# 03 - Model Experiments: 2-Layer LSTM

Builds the PyTorch LSTM architecture (80 hidden units, 0.3 dropout) and inspects its parameter profile.


```python
import sys
from pathlib import Path
import torch

PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.models.lstm_model import LSTMAnomalyDetector
from src.utils.config import load_config

cfg = load_config(str(PROJECT_ROOT / "config.yaml"))
n_features = 25  # Channel P-1 has 25 features

model = LSTMAnomalyDetector(
    input_size=n_features,
    hidden_size=80,
    num_layers=2,
    dropout=0.3
)
print(model)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total trainable parameters: {total_params:,}")
```

### Forward Pass Verification


```python
dummy_input = torch.randn(16, 250, n_features)
output = model(dummy_input)
print("Output shape:", output.shape)
assert output.shape == (16,)
print("Forward pass successful!")
```
